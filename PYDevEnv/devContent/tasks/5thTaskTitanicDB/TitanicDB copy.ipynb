## PREPROCESS

### Definición de variables explicativas y pipeline
Seleccionamos los atributos más informativos y aplicamos un preprocesamiento que resuelve valores faltantes, escala las columnas numéricas y codifica las categóricas para dejar la matriz lista para el PCA.

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
```

```python


feature_cols = ['Pclass', 'Sex', 'Age', 'SibSp', 'Parch', 'Fare', 'Embarked']
X = df[feature_cols].copy()
y = df['Survived'].copy()

numeric_features = ['Age', 'SibSp', 'Parch', 'Fare']
categorical_features = ['Sex', 'Embarked', 'Pclass']

def build_preprocessor():
    numeric_pipeline = Pipeline(
        steps=[
            ('imputer', SimpleImputer(strategy='median')),
            ('scaler', StandardScaler())
        ]
    )
    categorical_pipeline = Pipeline(
        steps=[
            ('imputer', SimpleImputer(strategy='most_frequent')),
            ('encoder', OneHotEncoder(handle_unknown='ignore'))
        ]
    )
    return ColumnTransformer(
        transformers=[
            ('numeric', numeric_pipeline, numeric_features),
            ('categorical', categorical_pipeline, categorical_features)
        ],
        sparse_threshold=0
    )

preprocessor = build_preprocessor()
X_preprocessed = preprocessor.fit_transform(X)
feature_names = preprocessor.get_feature_names_out()
X_preprocessed_df = pd.DataFrame(X_preprocessed, columns=feature_names)
X_preprocessed_df.head()
```

### PCA sobre las variables estandarizadas
Calculamos todos los componentes para medir cuánta varianza explica cada uno y definir un punto de corte razonable.

```python
from sklearn.decomposition import PCA
import numpy as np
import matplotlib.pyplot as plt 
```

```python

pca_full = PCA().fit(X_preprocessed)
explained_variance = pd.DataFrame(
    {
        'component': np.arange(1, len(pca_full.explained_variance_ratio_) + 1),
        'variance_ratio': pca_full.explained_variance_ratio_,
        'cumulative_variance': np.cumsum(pca_full.explained_variance_ratio_),
    }
)
explained_variance
```

```python
plt.figure(figsize=(8, 4))
plt.plot(
    explained_variance['component'],
    explained_variance['cumulative_variance'],
    marker='o'
)
plt.axhline(0.9, color='red', linestyle='--', label='90% varianza')
plt.xlabel('Número de componentes principales')
plt.ylabel('Varianza explicada acumulada')
plt.title('Curva de varianza explicada por PCA')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
component_names = [f'PC{i}' for i in range(1, pca_full.n_components_ + 1)]
loading_matrix = pd.DataFrame(
    pca_full.components_,
    columns=feature_names,
    index=component_names
)
loading_matrix.loc[['PC1', 'PC2', 'PC3']].T.sort_values(
    'PC1', key=lambda col: col.abs(), ascending=False
).head(10)
```

```python

pca_2 = PCA(n_components=2, random_state=42)
principal_components = pca_2.fit_transform(X_preprocessed)
principal_df = pd.DataFrame(principal_components, columns=['PC1', 'PC2'])
principal_df['Survived'] = y.values

plt.figure(figsize=(6, 5))
sns.scatterplot(
    data=principal_df,
    x='PC1',
    y='PC2',
    hue='Survived',
    palette='Set1',
    alpha=0.7
)
plt.title('Proyección de los pasajeros en los dos primeros componentes')
plt.tight_layout()
plt.show()
```

## MODELING

### Pipeline con PCA + Regresión Logística
Utilizamos el mismo preprocesamiento dentro de un pipeline de scikit-learn, reducimos la dimensionalidad hasta explicar ~90 % de la varianza y medimos el desempeño con validación cruzada estratificada.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import StratifiedKFold, cross_val_score

model_pipeline = Pipeline(
    steps=[
        ('preprocess', build_preprocessor()),
        ('pca', PCA(n_components=0.9, random_state=42)),
        ('log_reg', LogisticRegression(max_iter=2000))
    ]
)

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
scores = cross_val_score(model_pipeline, X, y, cv=cv, scoring='accuracy')

print(f'Accuracy media: {scores.mean():.3f}')
print(f'Desviación estándar: {scores.std():.3f}')
```

Con casi 80 % de acierto y variación baja entre pliegues (±2.5 pp), el modelo es estable y no depende de un solo split; el PCA no deteriora la capacidad predictiva y compacta la información sin sacrificar consistencia.

entrenamos nuevamente con las componentes para visualizar mediante un diagrama de dispersión pero en 3D

```python
from mpl_toolkits.mplot3d import Axes3D  # noqa: F401 (needed for 3D plots)

pca_3 = PCA(n_components=3, random_state=42)
principal_components_3 = pca_3.fit_transform(X_preprocessed)

principal_df_3d = pd.DataFrame(
    principal_components_3, columns=['PC1', 'PC2', 'PC3']
)
principal_df_3d['Survived'] = y.values

fig = plt.figure(figsize=(7, 6))
ax = fig.add_subplot(111, projection='3d')
scatter = ax.scatter(
    principal_df_3d['PC1'],
    principal_df_3d['PC2'],
    principal_df_3d['PC3'],
    c=principal_df_3d['Survived'],
    cmap='Set1',
    alpha=0.75
)
ax.set_xlabel('PC1'); ax.set_ylabel('PC2'); ax.set_zlabel('PC3')
ax.set_title('Titanic passengers in first 3 PCs')
fig.colorbar(scatter, ax=ax, label='Survived')
plt.tight_layout()
plt.show()
```

El “globo” rojo (supervivientes) se desplaza hacia valores positivos de PC1 y PC2, mientras que los puntos grises (no supervivientes) se concentran alrededor de PC1≈0 y PC2 bajos. 
Vemos que PC1 combina sobretodo Fare, Pclass y Sex, lo que confirma que el grupo con mayor poder adquisitivo y mujeres domina la zona roja.

No hay cúmulos aislados sino un gradiente: conforme se avanza por PC1/PC2, la probabilidad de supervivencia aumenta de forma continua. Eso sugiere que modelos lineales después del PCA pueden capturar bien esta tendencia.

tambien podemos examinar relaciones bivariadas entre los diferentes componetes

```python
import seaborn as sns

sns.pairplot(
    principal_df_3d[['PC1', 'PC2', 'PC3', 'Survived']],
    hue='Survived',
    diag_kind='kde',
    corner=True
)
plt.suptitle('Pairplot of principal components vs. survival', y=1.02)
plt.show()
```

En los histogramas diagonales, PC1 distingue mejor a los grupos, con los sobrevivientes desplazados a la derecha, mientras que PC2 y PC3 se solapan casi por completo. Los diagramas de dispersión confirman que la separación útil se concentra en PC1 y, en menor medida, en su combinación con PC2.

No se observan clusters definidos, por lo que el PCA muestra solo tendencias generales; para clasificar la supervivencia es preferible un modelo supervisado como la regresión logística.
